```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

from pandas.io.common import file_path_to_url

# This just makes tables print nicely and wide enough to read
pd.set_option("display.max_columns", 50)
pd.set_option("display.width", 140)

print("Libraries loaded successfully.")
```

```python
raw_data_folder = "data/raw"

iec_2016_file = os.path.join(raw_data_folder, "iEC_2016_ward_results.csv")
iec_2021_file = os.path.join(raw_data_folder, "iEC_2021_ward_results.csv")
census_file = os.path.join(raw_data_folder, "census2022_ward_population.csv")
crosswalk_file = os.path.join(raw_data_folder, "ward_crosswalk_2016_2021.csv")
for name, path in[("2016 election results", iec_2016_file),
                  ("2021 election results", iec_2021_file),
                  ("Census population data", census_file),
                  ("Ward Crosswalk", crosswalk_file)]:

    if os.path.exists(path):
        print(f"Found {name} {path}")
    else:
        print(f"MISSING {name} {path}")
```

```python
if os.path.exists(crosswalk_file):
    ward_crosswalk = pd.read_csv(crosswalk_file)
    print("Crosswalk loaded successfully. Here are the fist few rows:")
    print(ward_crosswalk.head())
else:
    ward_crosswalk = None
    print("Crosswalk not found. we will skip the 2016 to 2021")
```
